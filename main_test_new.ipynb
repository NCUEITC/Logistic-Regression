# 引入 Python 庫，`pandas`, `numpy`, `json`

````python
# 引入 `pandas`, `numpy`, `json`
import pandas as pd
import numpy as np
import json
````

### **第一行：`import pandas as pd`**
1. **`import` 是什麼？**
   - 在 Python 中，`import` 是用來引入其他模組（module）或庫（library）的關鍵字。模組可以包含事先寫好的功能和工具，讓我們不用從零開始實現所有功能。

2. **`pandas` 是什麼？**
   - `pandas` 是一個非常強大的 Python 資料分析工具，專門用來處理結構化資料（例如：表格、Excel 文件、資料庫資料）。
   - 它提供了兩種核心資料結構：
     - **DataFrame**：類似於 Excel 表格的結構，用來儲存表格式資料。
     - **Series**：類似於 Excel 中的一列或一行的資料。

3. **`as pd` 的作用**
   - 我們用 `as pd` 給 `pandas` 起一個簡短的別名。這樣在後面程式碼中，直接用 `pd` 呼叫它的功能，寫起來更方便。

**這行程式碼的總結：**
- 它引入了 `pandas` 模組，並且將其命名為 `pd`。
---

### **第二行：`import numpy as np`**
1. **`numpy` 是什麼？**
   - `numpy` 是一個用來處理數值運算的 Python 庫，尤其適合用於數學運算和處理多維陣列（矩陣）。
   - 它支援快速的矩陣運算（比純 Python 快很多），並提供許多數學和統計函數。

2. **`as np` 的作用**
   - 和第一行的 `as pd` 一樣，這裡把 `numpy` 簡稱為 `np`，方便後續使用。

**這行程式碼的總結：**
- 它引入了 `numpy` 模組，並將其命名為 `np`。
---

### **第三行：`import json`**
1. **`json` 是什麼？**
   - `json` 是 Python 的一個內建模組，用來處理 JSON 資料格式。
   - **JSON**（JavaScript Object Notation）是一種常見的資料交換格式，像是 `{"name": "Alice", "age": 25}`。很多 API 或檔案會使用 JSON 格式來傳遞資料。

2. **這行為什麼不需要 `as`？**
   - 這裡直接用 `import json`，說明程式碼作者認為直接使用 `json` 這個名字就足夠方便。

**這行程式碼的總結：**
- 它引入了處理 JSON 資料格式的內建模組。
---

### **整體概念總結：**
這三行程式碼引入了三個非常常用的 Python 庫，它們的用途如下：
- **`pandas`**：處理結構化的資料（表格）。
- **`numpy`**：進行高效的數值運算（數學、矩陣運算）。
- **`json`**：處理 JSON 資料格式，適用於讀取或寫入 API 資料。

## 用 `pandas` 讀取 csv 檔案

````python
# 變數 df 存取 .csv 檔案中的資料
csv_file_path = r"unprocessed_iris_dataset.csv"
df = pd.read_csv(csv_file_path)
````

### **第一行：`csv_file_path = r"unprocessed_iris_dataset.csv"`**
1. **`csv_file_path` 是什麼？**
   - 這是一個變數，用來儲存檔案路徑或檔案名稱。變數的名字是程式設計師自己定義的，在這裡它代表「CSV 檔案的路徑」。

2. **`r` 的作用**
   - `r` 表示「原始字串」（raw string）。
   - 在 Python 中，普通字串裡如果有反斜線（`\`），可能會被解讀為特殊字元（例如，`\n` 表示換行）。用 `r` 就告訴 Python，不要解讀字串中的反斜線，只是把它當作普通文字。
   - 例如：`r"C:\new_folder\file.csv"` 不會把 `\n` 解讀為換行。

3. **`"unprocessed_iris_dataset.csv"` 是什麼？**
   - 這是一個 CSV 檔案的名稱，表示存放在當前目錄下的檔案。
   - **CSV**（Comma-Separated Values）是一種用逗號分隔數據的檔案格式，常用來儲存結構化資料，像表格一樣。

**這行的總結：**
- 建立了一個變數 `csv_file_path`，其值為 `unprocessed_iris_dataset.csv`，表示我們稍後要讀取的檔案。
---

### **第二行：`df = pd.read_csv(csv_file_path)`**
1. **`pd.read_csv` 是什麼？**
   - 這是 `pandas` 中的一個方法，用來讀取 CSV 檔案並將其轉換成一個 **DataFrame**。
   - **DataFrame** 是 `pandas` 的核心資料結構，類似於 Excel 的工作表，它有行（row）和列（column），非常適合用來分析和操作資料。

2. **`csv_file_path` 作為參數**
   - `pd.read_csv` 方法需要知道要讀取的檔案路徑，這裡我們用變數 `csv_file_path` 傳遞檔案名稱。

3. **`df` 是什麼？**
   - `df` 是另一個變數，通常在處理資料分析時，程式設計師習慣用 `df` 表示「DataFrame」。
   - 讀取後，`df` 就會包含 CSV 檔案中的所有資料。

**這行的總結：**
- 使用 `pandas` 的 `read_csv` 方法讀取 CSV 檔案，並將其存入 `df` 變數中。現在 `df` 是一個結構化的表格資料。
---

### **整體概念總結：**
這段程式碼的作用是：
1. 定義了一個變數 `csv_file_path`，用來指向 CSV 檔案的名稱或路徑。
2. 使用 `pandas` 的 `read_csv` 方法將檔案讀取進程式，並將其內容存入一個表格資料結構（DataFrame）中，變數名稱是 `df`。
---

假設 CSV 檔案 `unprocessed_iris_dataset.csv` 的內容是這樣：

讀取後，`df` 的內容就會像這樣：
```

0           5.1          3.5           1.4          0.2  setosa

```

## `df.info()` 快速檢視

````python
df.info()
````

### **詳解 `df.info()`**

#### **1. 功能概述**
- `df.info()` 是 `pandas` 提供的快速檢視方法，用來了解資料的結構和屬性，適合於初步探索資料。
- 它輸出的內容包括：
  1. 資料的總行數和列數。
  2. 每個欄位的名稱、資料型態（如數值型 `float64` 或文字型 `object`）。
  3. 每個欄位的非空值數量（用於檢查是否有缺失值）。
  4. 整體記憶體使用量。

#### **2. 範例：假設 `df` 是 Iris 資料集的一部分**
```python
import pandas as pd

# 模擬建立一個小型的 DataFrame
data = {
    "sepal_length": [5.1, 4.9, 4.7],
    "sepal_width": [3.5, 3.0, 3.2],
    "petal_length": [1.4, 1.4, 1.3],
    "petal_width": [0.2, 0.2, 0.2],
    "species": ["setosa", "setosa", "setosa"]
}

df = pd.DataFrame(data)

# 使用 df.info()
df.info()
```

#### **3. 執行後輸出：**
```
<class 'pandas.core.frame.DataFrame'>
RangeIndex: 3 entries, 0 to 2
Data columns (total 5 columns):
 #   Column         Non-Null Count  Dtype  
---  ------         --------------  -----  
 0   sepal_length   3 non-null      float64
 1   sepal_width    3 non-null      float64
 2   petal_length   3 non-null      float64
 3   petal_width    3 non-null      float64
 4   species        3 non-null      object 
dtypes: float64(4), object(1)
memory usage: 248.0 bytes
```
---

#### **4. 輸出解讀**
1. **`<class 'pandas.core.frame.DataFrame'>`**
   - 這說明 `df` 是一個 `DataFrame` 資料結構。
2. **`RangeIndex: 3 entries, 0 to 2`**
   - `df` 共有 3 行，索引範圍是 0 到 2。
3. **`Data columns (total 5 columns):`**
   - `df` 總共有 5 個欄位。
4. **欄位資訊**
   - 每列包括：
     - 欄位名稱（`Column`）。
     - 非空值數量（`Non-Null Count`）。
     - 資料型態（`Dtype`）。
   - 例如，`sepal_length` 有 3 筆非空值，資料型態是 `float64`。
5. **`dtypes: float64(4), object(1)`**
   - 表示這個表格中有 4 個 `float64`（數值型欄位）和 1 個 `object`（文字型欄位）。
6. **`memory usage: 248.0 bytes`**
   - 資料表佔用的記憶體大小（以位元組為單位）。
---

### **`df.info()` 的實用性**
- **快速檢查資料品質：**
  - 非空值數量可以幫助你發現哪些欄位可能有遺失值。
- **確認資料型態：**
  - 可以確認數值型、文字型等資料是否正確，比如數值是否誤讀為文字型。
- **記憶體使用：**
  - 當處理大數據時，了解記憶體使用量有助於評估性能。
---

## `df.describe()` 快速生成數據統計摘要

````python
df.describe()
````

### **`df.describe()` 詳細講解**

#### **`df.describe()` 功能概述**
- `df.describe()` 是 `pandas` 提供的一個方法，用來快速生成數據的統計摘要，適用於數值型資料。
- 它的輸出包括：
  - **統計指標**（如：平均值、標準差、最小值、最大值等）。
  - **行（列）方向的數據總結**。
---

### **語法**
```python
df.describe()
```
---

### **`df.describe()` 輸出的主要統計指標**
1. **`count`**：每個欄位中非空值（非缺失值）的數量。
2. **`mean`**：欄位數值的平均值。
3. **`std`**：欄位數值的標準差（用於衡量數據分散程度）。
4. **`min`**：欄位中的最小值。
5. **`25%`（第一四分位數）**：數據中 25% 的值低於此值。
6. **`50%`（中位數或第二四分位數）**：數據中 50% 的值低於此值。
7. **`75%`（第三四分位數）**：數據中 75% 的值低於此值。
8. **`max`**：欄位中的最大值。
---

### **範例**

假設我們有一個簡單的 `DataFrame`：

```python
import pandas as pd

# 模擬資料
data = {
    "sepal_length": [5.1, 4.9, 4.7, 5.0],
    "sepal_width": [3.5, 3.0, 3.2, 3.6],
    "petal_length": [1.4, 1.4, 1.3, 1.5],
    "petal_width": [0.2, 0.2, 0.2, 0.3],
}

df = pd.DataFrame(data)

# 使用 df.describe()
print(df.describe())
```
---

#### **執行結果：**
```
       sepal_length  sepal_width  petal_length  petal_width
count      4.000000     4.000000      4.000000     4.000000
mean       4.925000     3.325000      1.400000     0.225000
std        0.170783     0.258844      0.083333     0.050000
min        4.700000     3.000000      1.300000     0.200000
25%        4.850000     3.150000      1.375000     0.200000
50%        4.950000     3.350000      1.400000     0.200000
75%        5.025000     3.525000      1.425000     0.250000
max        5.100000     3.600000      1.500000     0.300000
```
---

#### **輸出解讀**
1. **`count`**
   - 每個欄位有 4 個數值，無遺失值。
2. **`mean`**
   - 平均值，例如 `sepal_length` 平均為 `4.925`。
3. **`std`**
   - 標準差，例如 `sepal_width` 的標準差為 `0.258844`，表示數值波動不大。
4. **`min` 和 `max`**
   - 最小值和最大值，例如 `petal_width` 最小為 `0.2`，最大為 `0.3`。
5. **四分位數**
   - 用於描述數據分佈，例如 `sepal_length` 中有 25% 的值小於 `4.85`。
---

### **表格整理**

| **語法**           | **功效**                                                            | **實際範例**                                                                 |
|--------------------|--------------------------------------------------------------------|-------------------------------------------------------------------------------|
| `df.describe()`    | 提供數值型欄位的統計摘要，包括平均值、標準差、最大值、最小值等。                  | 查看資料分佈狀況，例如用於 Iris 數據集，總結各數值欄位的統計資訊。               |
| `df.describe(include='all')` | 將摘要擴展到非數值型欄位，例如類別型欄位的頻率和唯一值數量等。                    | `df.describe(include='all')` 顯示包括數值與類別欄位的摘要資訊。               |
| `df.describe(percentiles=[.1, .5, .9])` | 自定義輸出的百分位數，例如只顯示第 10%、50%、90% 的分佈資訊。              | `df.describe(percentiles=[.1, .5, .9])` 顯示自定義的百分位數統計摘要。         |
---

### **補充**
1. **數值型與非數值型欄位處理：**
   - 預設只會處理數值型欄位。若要包括文字或分類型欄位，需使用 `include='all'`。
2. **用途：**
   - 初步探索數據分佈（如是否有極端值）。
   - 驗證欄位是否有缺失值（藉由 `count` 和資料總行數比對）。
3. **與 `info()` 的區別：**
   - `info()` 側重於資料的結構（欄位名稱、型態、非空值數量）。
   - `describe()` 側重於統計摘要，主要是數值型資料。

## `df.head()` 快速檢視前幾行資料

````python
df.head()
````

### **`df.head()` 詳細講解**

#### **功能概述**
- `df.head()` 是 `pandas` 提供的一個方法，用來快速檢視 `DataFrame` 的前幾行資料。
- 預設會顯示前 5 行資料，你也可以透過參數自定義顯示的行數。
- 常用於探索資料集，快速確認數據格式和內容是否正確。
---

### **語法**
```python
df.head(n=5)
```

#### **參數**
- **`n`**：
  - 要顯示的行數，預設值是 5。
  - 你可以傳入任意正整數來改變顯示行數。
---

### **範例**

假設我們有一個簡單的 `DataFrame`：

```python
import pandas as pd

# 模擬數據
data = {
    "sepal_length": [5.1, 4.9, 4.7, 5.0, 5.4, 5.2],
    "sepal_width": [3.5, 3.0, 3.2, 3.6, 3.9, 3.4],
    "petal_length": [1.4, 1.4, 1.3, 1.5, 1.7, 1.6],
    "petal_width": [0.2, 0.2, 0.2, 0.3, 0.4, 0.3]
}

# 建立 DataFrame
df = pd.DataFrame(data)

# 使用 df.head()
print(df.head())
```
---

#### **執行結果**
```
   sepal_length  sepal_width  petal_length  petal_width
0           5.1          3.5           1.4          0.2
1           4.9          3.0           1.4          0.2
2           4.7          3.2           1.3          0.2
3           5.0          3.6           1.5          0.3
4           5.4          3.9           1.7          0.4
```
---

#### **詳細解讀**
1. **輸出格式**
   - `df.head()` 會顯示 `DataFrame` 的前 `n` 行，按照原始索引順序。
   - 在這裡，前 5 行資料（索引 0 到 4）被顯示出來。

2. **欄位名稱**
   - 第一行顯示 `DataFrame` 的欄位名稱，例如 `sepal_length`、`sepal_width` 等。

3. **數據內容**
   - 從每一列的數據中提取出前 `n` 行並顯示，例如 `sepal_length` 前 5 個值為 `5.1, 4.9, 4.7, 5.0, 5.4`。

4. **索引**
   - 左側的數字 `0, 1, 2, 3, 4` 是行索引，表示每一行在資料表中的唯一位置。
---

### **自定義顯示行數**

#### 顯示前 3 行資料：
```python
print(df.head(3))
```

##### **執行結果**
```
   sepal_length  sepal_width  petal_length  petal_width
0           5.1          3.5           1.4          0.2
1           4.9          3.0           1.4          0.2
2           4.7          3.2           1.3          0.2
```

#### 顯示前 10 行資料：
```python
print(df.head(10))
```
- 如果 `DataFrame` 的行數少於 10 行（例如只有 6 行），`head` 只會顯示全部資料，而不會報錯。
---

### **應用場合**
1. **快速檢查資料結構**
   - 確認欄位名稱是否正確。
   - 確保數據已正確載入到 `DataFrame`。

2. **查看數據內容**
   - 了解數據的大致分佈和內容格式。

3. **探索性資料分析**
   - 開始進行資料檢查和清理工作之前，快速檢視前幾行資料。

---

### **與其他方法的比較**
- **`df.tail()`**：
  - 與 `head()` 相反，用於顯示 `DataFrame` 的最後幾行。

## `df.duplicated()` 檢測是否存在重複的行

````python
df.duplicated()
````

### **`df.duplicated()` 詳細講解**

#### **功能概述**
- **`df.duplicated()`** 是 `pandas` 提供的一個方法，用來檢測 `DataFrame` 中是否存在重複的行。
- 返回的是一個布林值（`True/False`）的序列：
  - **`True`** 表示該行是重複的（與前面出現的某行一模一樣）。
  - **`False`** 表示該行是唯一的（未重複）。
- 此方法非常適合在資料清理時檢查和處理重複數據。
---

### **語法**
```python
df.duplicated(subset=None, keep='first')
```

#### **參數**
- **`subset`**：
  - 指定用來檢測重複的列（欄位）。如果未指定，將檢查所有列。
  - 例如：`subset=['col1', 'col2']`。
- **`keep`**：
  - 指定在檢測重複行時，保留哪一個版本：
    - **`'first'`（預設）**：標記除了第一個出現的重複行以外的所有重複行為 `True`。
    - **`'last'`**：標記除了最後一個出現的重複行以外的所有重複行為 `True`。
    - **`False`**：標記所有重複行為 `True`（包括第一個或最後一個）。

#### **返回值**
- 返回一個與 `DataFrame` 行數相同的布林值序列。

---

## 練習：找到並列出重複的行

````python
def print_duplicated_index(df):
    
    # 找到重複的索引
    duplicated_indices = df.index[df.duplicated()].unique()

    if len(duplicated_indices) > 0:
        print("Duplicated indices found:")
        for idx in duplicated_indices:
            print(f"\nIndex: {idx}")
            print(df.loc[idx])
    else:
        print("No duplicated indices found.")
print_duplicated_index(df)
````


  - `distinct_df` 中只會包含唯一的行。

#### **3. 重設索引**

```python
distinct_df = distinct_df.reset_index(drop=True)
```

  - 重設 `DataFrame` 的索引。
  
- **參數：**

- **用途：**
  - 在刪除重複行後，行索引可能會不連續（例如索引：0, 2, 5）。使用 `reset_index` 可以將索引重設為連續的。

## 使用 Pandas 繪製散佈圖

````python
"""使用 Pandas 繪製散佈圖，動態生成分群顏色，並標示極端值，圖例顯示欄位實際值。
:param df: pandas.DataFrame，包含數據的資料框
:param id_col: str，作為 X 軸的欄位名稱（如 ID）
:param group_col: str，作為分群的欄位名稱（如 Species）
:param lower_percentile: float，極端值的下百分比閾值
:param upper_percentile: float，極端值的上百分比閾值
"""
def plot_outliers_with_species_dynamic_colors(df, id_col, group_col, lower_percentile=0.01, upper_percentile=0.99):
    # 動態生成顏色映射
    unique_groups = df[group_col].unique()
    group_colors = {group: f"C{i}" for i, group in enumerate(unique_groups)}  # 使用 Pandas 支援的 C0, C1,... 顏色

    numeric_columns = df.select_dtypes(include='number').columns.drop(id_col)  # 選取數值型欄位，排除 ID

    for col in numeric_columns:
        # 計算百分位範圍
        lower_bound = df[col].quantile(lower_percentile)
        upper_bound = df[col].quantile(upper_percentile)

        # 檢測極端值
        df['Outlier'] = (df[col] <= lower_bound) | (df[col] >= upper_bound)

        # 初始化圖表
        ax = None
        
        # 按分群繪製非極端值
        for group in unique_groups:
            group_data = df[(df[group_col] == group) & (~df['Outlier'])]
            ax = group_data.plot.scatter(
                x=id_col,
                y=col,
                color=group_colors[group],  # 動態顏色
                label=group,  # 圖例顯示實際群組名稱
                figsize=(10, 6),
                ax=ax,
                title=f"Scatter Plot of {col} with Outliers Highlighted"
            )
        
        # 繪製極端值
        outliers = df[df['Outlier']]
        outliers.plot.scatter(
            x=id_col,
            y=col,
            color='red',
            label='Outliers',  # 極端值的圖例
            ax=ax
        )

        # 打印上下界資訊
        print(f"{col}: Lower Bound = {lower_bound:.2f}, Upper Bound = {upper_bound:.2f}")

        # 清理臨時欄位
        df.drop(columns=['Outlier'], inplace=True)

plot_outliers_with_species_dynamic_colors(no_Null_df, id_col='Id', group_col='Species', lower_percentile=0.01, upper_percentile=0.99)
````

## 移除資料框中指定數值欄位的極端值（基於百分位數閾值）。

````python
"""移除資料框中指定數值欄位的極端值（基於百分位數閾值）。

:param df: pandas.DataFrame，包含數據的資料框
:param numeric_columns: list，指定需要檢測極端值的數值型欄位名稱
:param lower_percentile: float，極端值的下百分比閾值（默認為 0.01）
:param upper_percentile: float，極端值的上百分比閾值（默認為 0.99）
:return: pandas.DataFrame，不含極端值的資料框
"""
def drop_outliers(df, numeric_columns, lower_percentile=0.01, upper_percentile=0.99):
    df_cleaned = df.copy()  # 創建副本，避免修改原始資料

    for col in numeric_columns:
        # 計算百分位範圍
        lower_bound = df[col].quantile(lower_percentile)
        upper_bound = df[col].quantile(upper_percentile)

        # 篩選非極端值
        df_cleaned = df_cleaned[(df_cleaned[col] > lower_bound) & (df_cleaned[col] < upper_bound)]

        # 打印篩選條件
        print(f"{col}: Lower Bound = {lower_bound:.2f}, Upper Bound = {upper_bound:.2f}, Remaining Rows = {len(df_cleaned)}")
    
    return df_cleaned

# 指定需要檢測極端值的數值型欄位
numeric_columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']

# 調用函式移除極端值
cleaned_df = drop_outliers(no_Null_df, numeric_columns, lower_percentile=0.01, upper_percentile=0.99)
````

# 二、損失函數 Loss

## RMSE：誤差評估指標

````python
def rmse(targets, predictions):
    return np.sqrt(np.mean(np.square(targets - predictions)))
````

### 定義 RMSE 函數
均方根誤差（Root Mean Square Error, RMSE）是一種常用於回歸模型的誤差評估指標，其公式如下：

$$
\text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^n (\text{targets}_i - \text{predictions}_i)^2}
$$


### 特性
1. **輸入參數**：
   - `targets`：實際值的數組或序列。
   - `predictions`：模型估算值的數組或序列。
2. **輸出**：
   - RMSE 值，用於衡量估算值與實際值之間的誤差大小。
3. **適用場景**：
   - 在回歸問題中評估模型的預測性能。
   - RMSE 的值越小，表示模型的估算值越接近實際值。

## 使用 RMSE 評估擬合效果

````python
target = iris_df.PetalLengthCm
w = 10
b = 100
estimated_PetalLengthCm = estimate_length(PetalWidthCm, w, b)
rmse(target, estimated_PetalLengthCm)
````

````python
def try_parameters(w, b):
    PetalWidthCm = iris_df.PetalWidthCm
    target = iris_df.PetalLengthCm

    estimated_PetalLengthCm = estimate_length(PetalWidthCm, w, b)

    plt.plot(PetalWidthCm, estimated_PetalLengthCm, 'r', alpha=0.9);
    plt.scatter(PetalWidthCm, target, s=8,alpha=0.8);
    plt.xlabel('PetalWidthCm');
    plt.ylabel('PetalLengthCm')
    plt.legend(['Estimate', 'Actual']);
    loss = rmse(target, estimated_PetalLengthCm)
    print("RMSE Loss: ", loss)

try_parameters(5, 10)
````
